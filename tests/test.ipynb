```python
import os
import h5py
```

```python
# load data
file_path = "../data/experiment_20250910-115100.h5"
with h5py.File(file_path, "r") as f:
    data = f['fly_data'][:]
n_animals = data.shape[1]
```

```python
import pandas as pd
import numpy as np

dfs = []
for name in data.dtype.names:
    print(f"{name}, {data.dtype[name]}")
    sub_df = pd.DataFrame(data[name])
    sub_df = sub_df.melt(var_name='animal', value_name=name).reset_index()
    dfs.append(sub_df)
    print(sub_df.shape)

# merge the dataframes by the index and animal value
for i in range(1, len(dfs)):
    df = pd.merge(df, dfs[i], on=['index', 'animal'], suffixes=('', f'_{i+1}'))
```
